```python
data = {
    "temperature": [18, 21, 24, 27, 30],
    "humidity": [75, 70, 65, 60, 55],
    "pollution": [180, 165, 140, 115, 95]
}

df = pd.DataFrame(data)

df
```

```python
#first visualization
```

```python
plt.figure(figsize=(8, 5))

plt.plot(
    df["temperature"],
    df["pollution"],
    marker="o"
)

plt.xlabel("Temperature")
plt.ylabel("Pollution")
plt.title("Temperature vs Pollution")

plt.show()
```
